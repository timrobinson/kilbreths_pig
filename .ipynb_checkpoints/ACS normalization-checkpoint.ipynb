```python
# If needed:
# pip install pymupdf pandas requests rapidfuzz python-dateutil

import re
import time
import json
import requests
import pandas as pd

from rapidfuzz import fuzz
from dateutil import parser as dateparser

import fitz  # PyMuPDF
```

```python
REFS_HEADER_RE = re.compile(r"^\s*references\s*$", re.IGNORECASE)
BRACKET_REF_RE = re.compile(r"^\s*\[\s*(\d+)\s*\]\s*(.*)$")
DOT_REF_RE     = re.compile(r"^\s*(\d+)\.\s+(.*)$")

def extract_pdf_text_by_page(pdf_path: str):
    doc = fitz.open(pdf_path)
    pages = []
    for i in range(len(doc)):
        pages.append(doc[i].get_text("text"))
    return pages

def find_references_start_page(pages):
    """
    Find the first page that contains a standalone 'References' line.
    Fallback: first page with 'References' anywhere.
    """
    for i, t in enumerate(pages):
        lines = [ln.strip() for ln in t.splitlines()]
        if any(REFS_HEADER_RE.match(ln) for ln in lines):
            return i
    for i, t in enumerate(pages):
        if re.search(r"\breferences\b", t, re.IGNORECASE):
            return i
    return None

def slice_references_text(pages, start_page):
    """Take text from start_page to end."""
    if start_page is None:
        return ""
    return "\n".join(pages[start_page:])

def split_numbered_references(ref_text: str):
    """
    Split into ordered (ref_id -> raw) using either [n] or 'n.'.
    Prefers bracketed ACS-style splitting.
    """
    lines = ref_text.splitlines()

    # Build a stream of "start markers"
    starts = []
    for idx, ln in enumerate(lines):
        m1 = BRACKET_REF_RE.match(ln)
        if m1:
            starts.append((idx, int(m1.group(1)), "bracket"))
            continue
        m2 = DOT_REF_RE.match(ln)
        if m2:
            starts.append((idx, int(m2.group(1)), "dot"))

    if not starts:
        return {}

    # Choose mode by majority in first ~200 markers
    sample = starts[:200]
    bracket_n = sum(1 for _, _, mode in sample if mode == "bracket")
    dot_n     = sum(1 for _, _, mode in sample if mode == "dot")
    mode = "bracket" if bracket_n >= dot_n else "dot"

    starts = [(i, n, m) for (i, n, m) in starts if m == mode]
    starts.sort(key=lambda x: x[0])

    entries = {}
    for k, (line_idx, n, _) in enumerate(starts):
        end = starts[k+1][0] if k+1 < len(starts) else len(lines)
        chunk_lines = lines[line_idx:end]

        # Remove the leading "[n]" or "n."
        first = chunk_lines[0]
        if mode == "bracket":
            m = BRACKET_REF_RE.match(first)
            lead = m.group(2) if m else first
        else:
            m = DOT_REF_RE.match(first)
            lead = m.group(2) if m else first

        chunk_lines[0] = lead
        raw = " ".join(x.strip() for x in chunk_lines if x.strip())
        entries[n] = raw.strip()

    return entries

def build_df_original_from_acs_pdf(pdf_path: str):
    pages = extract_pdf_text_by_page(pdf_path)
    start_page = find_references_start_page(pages)
    ref_text = slice_references_text(pages, start_page)
    entries = split_numbered_references(ref_text)

    df = pd.DataFrame(
        [{"ref_id": k, "raw": v} for k, v in sorted(entries.items(), key=lambda x: x[0])]
    )
    return df, {"references_start_page": start_page, "n_refs": len(df)}

# Example:
# df_original, report = build_df_original_from_acs_pdf("test.pdf")
# display(report); df_original.head()
```

```python
DOI_RE = re.compile(r"(10\.\d{4,9}/[^\s\"\'<>]+)", re.IGNORECASE)

def normalize_doi(doi: str) -> str:
    doi = (doi or "").strip()
    doi = doi.rstrip(").,;]}")
    doi = doi.replace("doi:", "").strip()
    return doi.lower()

def extract_doi_from_raw(raw: str):
    if not isinstance(raw, str):
        return None
    m = DOI_RE.search(raw)
    if not m:
        return None
    return normalize_doi(m.group(1))

def add_doi_column(df_original: pd.DataFrame) -> pd.DataFrame:
    df = df_original.copy()
    df["doi"] = df["raw"].apply(extract_doi_from_raw)
    df["doi_url"] = df["doi"].apply(lambda d: f"https://doi.org/{d}" if isinstance(d, str) and d else None)
    return df

# df_original = add_doi_column(df_original)
```

```python
CROSSREF_WORKS = "https://api.crossref.org/works"
OPENALEX_WORKS = "https://api.openalex.org/works"

def crossref_find_doi_from_citation(raw: str, timeout=20):
    """
    Crossref has a /works endpoint with query.bibliographic.
    Not perfect, but decent for ACS-like refs.
    """
    if not raw or not isinstance(raw, str):
        return None

    params = {
        "query.bibliographic": raw,
        "rows": 3,
    }
    r = requests.get(CROSSREF_WORKS, params=params, timeout=timeout, headers={"User-Agent": "kilbreths-pig/0.1 (mailto:you@example.com)"})
    if r.status_code != 200:
        return None

    data = r.json()
    items = data.get("message", {}).get("items", []) or []
    for it in items:
        doi = it.get("DOI")
        if doi:
            return normalize_doi(doi)
    return None

def openalex_find_doi_from_citation(raw: str, timeout=20):
    """
    OpenAlex supports a free-text 'search' parameter.
    """
    if not raw or not isinstance(raw, str):
        return None

    params = {"search": raw, "per-page": 3}
    r = requests.get(OPENALEX_WORKS, params=params, timeout=timeout, headers={"User-Agent": "kilbreths-pig/0.1"})
    if r.status_code != 200:
        return None

    data = r.json()
    results = data.get("results", []) or []
    for it in results:
        doi = (it.get("doi") or "").replace("https://doi.org/", "")
        if doi:
            return normalize_doi(doi)
    return None

def enrich_missing_dois(df_with_doi: pd.DataFrame, sleep_s=0.15) -> pd.DataFrame:
    df = df_with_doi.copy()
    found = []
    for _, row in df.iterrows():
        if row.get("doi"):
            found.append(row["doi"])
            continue

        raw = row.get("raw", "")
        doi = crossref_find_doi_from_citation(raw)
        if not doi:
            doi = openalex_find_doi_from_citation(raw)

        found.append(doi)
        time.sleep(sleep_s)

    df["doi_found"] = found
    df["doi_final"] = df["doi"].fillna(df["doi_found"])
    df["doi_final"] = df["doi_final"].apply(lambda d: normalize_doi(d) if isinstance(d, str) and d else None)
    df["doi_status"] = df["doi_final"].apply(lambda d: "OK" if d else "No DOI")
    df["doi_url_final"] = df["doi_final"].apply(lambda d: f"https://doi.org/{d}" if d else None)
    return df

# df_original = enrich_missing_dois(df_original)
```

```python
def _safe_join(xs, sep="; "):
    xs = [x for x in (xs or []) if x]
    return sep.join(xs) if xs else None

def crossref_metadata(doi: str, timeout=20):
    if not doi:
        return None
    url = f"{CROSSREF_WORKS}/{doi}"
    r = requests.get(url, timeout=timeout, headers={"User-Agent": "kilbreths-pig/0.1 (mailto:you@example.com)"})
    if r.status_code != 200:
        return None
    return r.json().get("message", {})

def crossref_to_acs_fields(msg: dict):
    if not msg:
        return {}

    title = (msg.get("title") or [None])[0]
    container = (msg.get("container-title") or [None])[0]  # journal
    issued = msg.get("issued", {}).get("date-parts", [[None]])
    year = issued[0][0] if issued and issued[0] else None

    authors = msg.get("author") or []
    # ACS author string: "Last, F. M.; Last, F. M.; ..."
    author_tokens = []
    for a in authors:
        family = a.get("family")
        given = a.get("given")
        if family and given:
            initials = " ".join([p[0] + "." for p in re.split(r"[\s-]+", given.strip()) if p])
            author_tokens.append(f"{family}, {initials}")
        elif family:
            author_tokens.append(family)
    authors_acs = _safe_join(author_tokens, sep="; ")

    return {
        "doi": normalize_doi(msg.get("DOI")),
        "doi_url": f"https://doi.org/{normalize_doi(msg.get('DOI'))}" if msg.get("DOI") else None,
        "title": title,
        "authors": authors_acs,
        "journal": container,
        "year": year,
        "volume": msg.get("volume"),
        "issue": msg.get("issue"),
        "pages": msg.get("page"),
        "publisher": msg.get("publisher"),
        "type": msg.get("type"),
    }

def build_df_doi(df_original: pd.DataFrame, sleep_s=0.15) -> pd.DataFrame:
    rows = []
    for _, row in df_original.iterrows():
        doi = row.get("doi_final")
        if not doi:
            rows.append({"ref_id": row.get("ref_id"), "doi": None})
            continue

        msg = crossref_metadata(doi)
        fields = crossref_to_acs_fields(msg)
        fields["ref_id"] = row.get("ref_id")
        rows.append(fields)
        time.sleep(sleep_s)

    return pd.DataFrame(rows)

# df_doi = build_df_doi(df_original)
# df_doi.head()
```

```python
def token_score(a, b):
    if not a or not b:
        return None
    return fuzz.token_set_ratio(str(a), str(b))

def partial_score(a, b):
    if not a or not b:
        return None
    return fuzz.partial_ratio(str(a), str(b))

def compute_match_scores(df_original: pd.DataFrame, df_doi: pd.DataFrame) -> pd.DataFrame:
    left = df_original.copy()
    right = df_doi.copy()

    merged = left.merge(
        right[["ref_id", "doi", "title", "authors", "journal", "year", "volume", "pages"]],
        on="ref_id",
        how="left",
        suffixes=("_pdf", "_doi"),
    )

    # Scores:
    merged["score_title"] = merged.apply(lambda r: token_score(r.get("raw"), r.get("title")), axis=1)
    merged["score_authors"] = merged.apply(lambda r: partial_score(r.get("raw"), r.get("authors")), axis=1)
    merged["score_journal"] = merged.apply(lambda r: partial_score(r.get("raw"), r.get("journal")), axis=1)
    merged["score_year"] = merged.apply(lambda r: partial_score(r.get("raw"), str(r.get("year")) if r.get("year") else None), axis=1)

    # Weighted overall score (tweak weights as you like)
    def overall(r):
        parts = []
        weights = []
        for key, w in [("score_title", 0.45), ("score_authors", 0.25), ("score_journal", 0.20), ("score_year", 0.10)]:
            v = r.get(key)
            if v is not None:
                parts.append(v * w)
                weights.append(w)
        if not weights:
            return None
        return sum(parts) / sum(weights)

    merged["match_score"] = merged.apply(overall, axis=1)

    # A simple status label
    def label(r):
        if r.get("doi_status") == "No DOI":
            return "No DOI"
        s = r.get("match_score")
        if s is None:
            return "Unknown"
        if s >= 90:
            return "Strong"
        if s >= 75:
            return "Moderate"
        return "Weak"

    merged["match_label"] = merged.apply(label, axis=1)
    return merged

# df_compare = compute_match_scores(df_original, df_doi)
# df_compare[["ref_id","doi_status","doi_final","match_score","match_label"]].head(10)
```
