# PDF bibliography → DataFrame (deterministic + LLM patch scaffold)
This notebook extracts numbered references from a PDF (ACS-style `(n)`), builds a DataFrame compatible with your Word pipeline, and optionally uses an LLM to patch missing/blank references.


```python
import re
import pandas as pd
import pdfplumber

try:
    import fitz  # PyMuPDF
except Exception:
    fitz = None

from dataclasses import dataclass, field
from typing import Optional, Dict, Any, List, Tuple
```

```python
@dataclass
class RefSpan:
    n: int
    raw: str                     # raw chunk (bounded candidate)
    det_conf: float = 0.0        # deterministic confidence score
    notes: List[str] = field(default_factory=list)
    # Optional provenance hooks (add later if you want)
    start_idx: Optional[int] = None
    end_idx: Optional[int] = None

def as_text(x) -> str:
    """Return plain text from either RefSpan or string."""
    if isinstance(x, RefSpan):
        return x.raw or ""
    return x or ""
```

```python
def pdf_to_text(pdf_path: str, method: str = "pdfplumber") -> str:
    """Extract text from a text-native PDF."""
    if method == "pdfplumber":
        chunks = []
        with pdfplumber.open(pdf_path) as pdf:
            for page in pdf.pages:
                txt = page.extract_text() or ""
                txt = re.sub(r"\r\n?", "\n", txt)
                chunks.append(txt)
        return "\n\n".join(chunks)

    if method == "pymupdf":
        if fitz is None:
            raise RuntimeError("PyMuPDF not available")
        doc = fitz.open(pdf_path)
        return "\n\n".join(page.get_text("text") or "" for page in doc)

    raise ValueError(f"Unknown method: {method}")
```

```python
def extract_references_section(text: str) -> str:
    """
    Return (mostly) the references list.

    Strategy:
    1) Prefer a true section heading at line start: "REFERENCES"
    2) From there, return the remainder (we stop later if needed).
    3) If no heading found, fallback to first "(1)" occurrence.
    """
    t = text.replace("\r\n", "\n").replace("\r", "\n")

    # 1) Strong signal: REFERENCES as a standalone heading line
    m = re.search(r"(?mi)^\s*references\s*$", t)
    if m:
        return t[m.end():].strip()

    # 2) Weaker: line-start "References" heading with optional punctuation
    m = re.search(r"(?mi)^\s*references\b\s*[:.]?\s*$", t)
    if m:
        return t[m.end():].strip()

    # 3) Fallback: start at first actual reference marker
    m = re.search(r"\(\s*1\s*\)", t)
    if m:
        return t[m.start():].strip()

    raise ValueError("Could not locate references section (no heading and no '(1)' marker).")
```

```python
def trim_to_first_reference(refs_text: str) -> str:
    """Trim everything before the first *true* reference start."""
    t = refs_text.replace("\r\n", "\n").replace("\r", "\n")
    m = re.search(r"\(\s*1\s*\)\s*[A-Z][A-Za-z'’\-]+\s*,", t)
    if m:
        return t[m.start():].strip()
    m = re.search(r"\(\s*1\s*\)", t)
    if m:
        return t[m.start():].strip()
    raise ValueError("Couldn't find a '(1)' marker in references text.")
```

```python
def normalize_for_splitting(text: str) -> str:
    """Keep newlines; only safe repairs."""
    t = text.replace("\r\n", "\n").replace("\r", "\n")
    t = re.sub(r"(\w)-\n(\w)", r"\1\2", t)   # join hyphenated line breaks
    t = re.sub(r"[ \t]+", " ", t)               # collapse spaces/tabs
    return t.strip()
```

```python
def normalize_entry(entry: str) -> str:
    """Flatten newlines within an entry."""
    t = entry.replace("\r\n", "\n").replace("\r", "\n")
    t = re.sub(r"\s*\n\s*", " ", t)
    t = re.sub(r"[ \t]+", " ", t)
    return t.strip()
```

```python
def ref_number(entry: str):
    m = re.match(r"^\(\s*(\d+)\s*\)", entry.strip())
    return int(m.group(1)) if m else None

def dedupe_keep_longest(entries: list[str]) -> list[str]:
    """If we over-split, keep the longest chunk for each ref number."""
    buckets = {}
    for e in entries:
        n = ref_number(e)
        if n is None:
            continue
        buckets.setdefault(n, []).append(e)
    best = {n: max(lst, key=len) for n, lst in buckets.items()}
    return [best[n] for n in sorted(best)]
```

```python
def strip_embedded_other_markers(entry: str) -> str:
    """If another '(k)' appears later in the same chunk, cut before it."""
    n = ref_number(entry)
    if n is None:
        return entry.strip()
    for m in re.finditer(r"\(\s*(\d+)\s*\)", entry):
        k = int(m.group(1))
        if k != n and m.start() > 0:
            return entry[:m.start()].strip()
    return entry.strip()
```

```python
def strip_comma_after_marker(s: str) -> str:
    return re.sub(r"^(\(\s*\d+\s*\)|\[\s*\d+\s*\])\s*,\s*", r"\1 ", (s or "").strip())
```

```python
_TRUNC_HEADINGS = re.compile(
    r"(?i)\b(AUTHOR INFORMATION|ACKNOWLEDGMENTS|Funding|Notes|Complete contact information is available)\b"
)

_SPAN_PATTERNS = [
    r"\bSupporting Information is available.*?(?=(\(\s*\d+\s*\))|$)",
    r"\bAdditional experimental details.*?(?=(\(\s*\d+\s*\))|$)",
    r"\bFigure\s*S\d+\b",
    r"\bFigures\s*S\d+(?:[–-]S\d+)?\b",
    r"\bTable\s*S\d+\b",
    r"\bTables\s*S\d+(?:\s*,\s*S\d+)*\b",
    r"\(PDF\)",
    r"\borcid\.org/\S+",
    r"\bEmail:\s*\S+",
]
_span_re = re.compile("|".join(_SPAN_PATTERNS), re.IGNORECASE)

def drop_leading_title_fragment(s: str) -> str:
    """
    If after the author list we start with a dangling lowercase fragment like
    'parative,' or 'ical' etc, drop up to the next sentence boundary.
    Conservative: only triggers if string starts with '('n') <authors> <lowercase...'
    """
    if not s:
        return s

    # keep marker
    m = re.match(r"^\s*(\(\s*\d+\s*\))\s*(.*)$", s)
    if not m:
        return s

    lead = m.group(1) + " "
    rest = m.group(2).lstrip()

    # If rest starts with lowercase letter and there's a comma soon, drop first token
    if re.match(r"^[a-z]{3,}", rest):
        # drop first token up to first space
        parts = rest.split(" ", 1)
        if len(parts) == 2:
            return (lead + parts[1].lstrip()).strip()

    return s


def clean_entry_det(entry: str) -> str:
    """Deterministic cleanup. IMPORTANT: should not erase the entire ref."""
    t = entry.replace("\r\n", "\n").replace("\r", "\n")

    m = _TRUNC_HEADINGS.search(t)
    if m:
        t = t[:m.start()]

    t = _span_re.sub(" ", t)
    t = normalize_entry(t)
    return t
```

```python
def strip_paren_artifacts(s: str) -> str:
    s = re.sub(r"\(\s*\)", "", s)                 # remove ( )
    s = re.sub(r"\(\s*[–-]\s*S\d+\s*\)", "", s)   # remove (−S5) style
    s = re.sub(r"\(\s*;\s*\)", "", s)             # rare weird case
    s = re.sub(r"\s{2,}", " ", s).strip()
    return s
```

```python
def ensure_marker(n: int, s: str) -> str:
    s = (s or "").strip()
    # Remove any leading marker-like token and replace with correct one
    s = re.sub(r"^\s*(\(\s*\d+\s*\)|\[\s*\d+\s*\]|\d+\s*[\.\)])\s*", "", s)
    return f"({n}) " + s.strip() if s else f"({n})"

```

```python
def _marker_plausible(refs: str, start: int, end: int) -> bool:
    window = refs[end:end+600]

    # Reject obvious author/affiliation boilerplate ONLY if it appears very early
    # (prevents rejecting legitimate citations that mention "University" in titles later)
    bad_early = re.search(
        r"(?i)\b(author information|corresponding author|department of|university|e-?mail:|orcid)\b",
        window[:180]
    )
    if bad_early:
        return False

    # Positive evidence (any one)
    has_year = re.search(r"\b(19|20)\d{2}\b", window) is not None
    has_doi  = re.search(r"\bdoi\b|10\.\d{4,9}/", window, re.I) is not None
    has_pages = re.search(r"\b\d{2,6}\s*[−-]\s*\d{2,6}\b", window[:140]) is not None
    has_author_punct = ("," in window[:260]) or (";" in window[:260])

    return has_year or has_doi or has_pages or has_author_punct

```

```python
def _find_next_marker(refs: str, n: int, cursor: int, window: int = 20000) -> tuple[int, int] | None:
    patterns = [
        rf"\(\s*{n}\s*\)",
        rf"\[\s*{n}\s*\]",
        rf"(?m)^\s*{n}\s*[\.\)]\s*",
    ]

    # Search in a bounded window first (prevents catastrophic cursor jumps)
    hi = min(len(refs), cursor + window)
    region = refs[cursor:hi]

    best = None
    for pat in patterns:
        m = re.search(pat, region)
        if not m:
            continue
        start = cursor + m.start()
        end = cursor + m.end()
        if best is None or start < best[0]:
            best = (start, end)

    if best is not None:
        return best

    # Fallback: global search after cursor (coverage recovery)
    for pat in patterns:
        m = re.search(pat, refs[cursor:])
        if m:
            return (cursor + m.start(), cursor + m.end())

    return None




def find_markers_monotonic(refs: str, expected_n: int) -> dict[int, tuple[int, int] | None]:
    pos: dict[int, tuple[int, int] | None] = {}
    cursor = 0

    MAX_JUMP = 30000  # chars; tune if needed

    for n in range(1, expected_n + 1):
        hit = _find_next_marker(refs, n, cursor, window=20000)  # <-- changed call

        # Guardrail: if the marker is absurdly far away, treat as missing
        if hit is not None and (hit[0] - cursor) > MAX_JUMP:
            hit = None

        # Skip obvious author/affiliation boilerplate occurrences
        while hit is not None:
            start, end = hit
            window = refs[end:end+250]
            if re.search(r"(?i)\b(author information|authors?\b|corresponding author|department of|university|orcid|e-?mail)\b",
                         window[:160]):
                cursor = end  # advance and try to find the NEXT occurrence of the same (n)
                hit = _find_next_marker(refs, n, cursor)
            else:
                break
                
        if hit is None:
            pos[n] = None
            continue

        pos[n] = hit
        cursor = hit[1]   # enforce monotonicity

    return pos

def truncate_at_embedded_marker_with_pos(chunk: str) -> tuple[str, bool, int | None]:
    if not chunk:
        return chunk, False, None

    lead = re.match(r"^\s*(\(\s*\d+\s*\)|\[\s*\d+\s*\]|\d+\s*[\.\)])\s*", chunk)
    search_start = lead.end() if lead else 0
    tail = chunk[search_start:]

    pat = re.compile(
        r"\(\s*\d+\s*\)|\[\s*\d+\s*\]|^\s*\d+\s*[\.\)]\s*",
        flags=re.MULTILINE
    )

    m = pat.search(tail)
    if not m:
        return chunk, False, None

    cut = search_start + m.start()
    return chunk[:cut].rstrip(), True, cut

def looks_like_dangling_stub(s: str) -> bool:
    if not s:
        return True
    if re.search(r"[,:;]\s*$", s):
        return True
    body = re.sub(r"^\(\s*\d+\s*\)\s*", "", s).strip()
    if len(body) < 80:
        return True
    return False

_BAD = re.compile(r"(?i)\b(author information|authors?\b|corresponding author|department of|university|orcid|e-?mail)\b")

def find_markers_all(refs: str, expected_n: int) -> dict[int, list[tuple[int, int]]]:
    """
    Collect ALL occurrences of reference markers for each n.
    Returns: {n: [(start,end), ...]}
    """
    occ: dict[int, list[tuple[int, int]]] = {n: [] for n in range(1, expected_n + 1)}

    # (n)
    for m in re.finditer(r"\(\s*(\d{1,3})\s*\)", refs):
        n = int(m.group(1))
        if 1 <= n <= expected_n:
            occ[n].append((m.start(), m.end()))

    # [n]
    for m in re.finditer(r"\[\s*(\d{1,3})\s*\]", refs):
        n = int(m.group(1))
        if 1 <= n <= expected_n:
            occ[n].append((m.start(), m.end()))

    # line-start n. / n)
    for m in re.finditer(r"(?m)^\s*(\d{1,3})\s*[\.\)]\s*", refs):
        n = int(m.group(1))
        if 1 <= n <= expected_n:
            occ[n].append((m.start(), m.end()))

    # sort & de-dup
    for n in occ:
        occ[n] = sorted(set(occ[n]), key=lambda x: x[0])

    return occ

def choose_best_marker(refs: str, candidates: list[tuple[int, int]]) -> tuple[int, int] | None:
    """
    Pick one marker occurrence for n.
    Prefer candidates NOT followed by obvious author/affiliation boilerplate.
    """
    if not candidates:
        return None

    for (s, e) in candidates:
        if _BAD.search(refs[e:e+160]):
            continue
        return (s, e)

    return candidates[0]  # fallback: earliest occurrence

def select_markers_by_number(refs: str, expected_n: int) -> dict[int, tuple[int, int] | None]:
    """
    For each n=1..expected_n, pick a single best marker occurrence globally.
    """
    occ = find_markers_all(refs, expected_n)
    return {n: choose_best_marker(refs, occ.get(n, [])) for n in range(1, expected_n + 1)}

def has_reference_payload(s: str) -> bool:
    # not semantic correctness—just “reference-ish payload”
    return (
        bool(re.search(r"\b(19|20)\d{2}\b", s)) or
        bool(re.search(r"\bdoi\b|10\.\d{4,9}/", s, re.I)) or
        bool(re.search(r"\bJ\.\b|\bProc\.\b|\bACS\b|\bNature\b|\bScience\b|\bPolymer\b", s, re.I))
    )


```

```python
def build_all_marker_starts(refs: str, expected_n: int) -> list[int]:
    starts = set()

    # (n)
    for m in re.finditer(r"\(\s*(\d{1,3})\s*\)", refs):
        n = int(m.group(1))
        if 1 <= n <= expected_n:
            if _marker_plausible(refs, m.start(), m.end()):
                starts.add(m.start())

    # [n]
    for m in re.finditer(r"\[\s*(\d{1,3})\s*\]", refs):
        n = int(m.group(1))
        if 1 <= n <= expected_n:
            if _marker_plausible(refs, m.start(), m.end()):
                starts.add(m.start())

    # line-start n. / n)
    for m in re.finditer(r"(?m)^\s*(\d{1,3})\s*[\.\)]\s*", refs):
        n = int(m.group(1))
        if 1 <= n <= expected_n:
            if _marker_plausible(refs, m.start(), m.end()):
                starts.add(m.start())

    return sorted(starts)


def next_marker_start_after(starts: list[int], start_n: int) -> int | None:
    """Return the next marker start strictly after start_n, else None."""
    # binary search
    import bisect
    i = bisect.bisect_right(starts, start_n)
    return starts[i] if i < len(starts) else None
```

```python
def truncate_at_second_author_block(s: str, min_cluster: int = 3, span: int = 120) -> str:
    """
    If s contains two distinct *clusters* of author tokens, truncate at the start
    of the second cluster.

    Cluster definition:
      - at least `min_cluster` occurrences of "Surname, X.;" within `span` chars.
    This avoids triggering on single tokens like "Hendrix, P."
    """
    if not s:
        return s

    # Remove leading marker for scanning
    lead_m = re.match(r"^\s*(\(\s*\d+\s*\))\s*", s)
    lead = (lead_m.group(1) + " ") if lead_m else ""
    body = re.sub(r"^\(\s*\d+\s*\)\s*", "", s).strip()

    # Token that looks like an author entry: "Surname, X.;"
    tok = re.compile(r"\b[A-Z][A-Za-z'`-]+,\s*[A-Z](?:\.-?[A-Z]\.)?\s*;")

    hits = [m.start() for m in tok.finditer(body)]
    if len(hits) < (2 * min_cluster):
        return s  # can't have two clusters

    # Build clusters: consecutive hits within `span`
    clusters = []
    i = 0
    while i < len(hits):
        j = i
        while j + 1 < len(hits) and (hits[j + 1] - hits[i]) <= span:
            j += 1
        count = j - i + 1
        if count >= min_cluster:
            clusters.append((hits[i], count))
        i += 1

    if len(clusters) < 2:
        return s

    # Truncate at the start of the SECOND cluster (earliest second cluster after the first)
    clusters = sorted(clusters, key=lambda x: x[0])
    cut = clusters[1][0]
    return (lead + body[:cut].rstrip()).rstrip()

```

```python
def truncate_at_and_followed_by_author_list(s: str, debug: bool = False) -> str:
    if not s:
        return s

    lead_m = re.match(r"^\s*(\(\s*\d+\s*\))\s*", s)
    lead = (lead_m.group(1) + " ") if lead_m else ""
    body = re.sub(r"^\(\s*\d+\s*\)\s*", "", s).strip()

    # Author token like: "MacKenzie,D.;" or "Taroc,A.-M.;" or "Smith, J;"
    author_tok = r"[A-Z][A-Za-z'`-]+,\s*[A-Z](?:\.[A-Z])?(?:\.-?[A-Z])?(?:\.)?\s*;"
    pat = re.compile(rf"\band\s+({author_tok})", re.IGNORECASE)

    m = pat.search(body)
    if not m:
        if debug:
            print("NO MATCH for and+author_tok")
        return s

    cut = m.start()  # at 'and ...'
    if debug:
        print("CUTTING at:", body[max(0, cut-30):cut+30])

    # avoid silly early cuts
    if cut < 40:
        return s

    return (lead + body[:cut].rstrip()).rstrip()

```

```python
def _payload_score(chunk: str) -> int:
    """
    Cheap deterministic 'reference-likeness' score.
    Higher is better.
    """
    if not chunk:
        return 0
    body = re.sub(r"^\s*(\(\s*\d+\s*\)|\[\s*\d+\s*\]|\d+\s*[\.\)])\s*", "", chunk).strip()

    score = 0
    score += min(len(body), 400)                  # length helps, capped
    score += 120 if re.search(r"\b(19|20)\d{2}\b", body) else 0
    score += 120 if re.search(r"\bdoi\b|10\.\d{4,9}/", body, re.I) else 0
    score += 60  if re.search(r"\b[A-Z][A-Za-z'`-]+,\s*[A-Z]", body) else 0  # looks like author token
    score -= 200 if _BAD.search(body[:180]) else 0                           # author-info boilerplate penalty
    score -= 250 if len(body) < 20 else 0
    return score

def select_markers_by_number_scored(refs: str, expected_n: int) -> dict[int, tuple[int,int] | None]:
    occ = find_markers_all(refs, expected_n)
    all_starts = build_all_marker_starts(refs, expected_n)

    chosen: dict[int, tuple[int,int] | None] = {}

    for n in range(1, expected_n + 1):
        cands = occ.get(n, [])
        if not cands:
            chosen[n] = None
            continue

        best = None
        best_score = -10**9

        for (s, e) in cands:
            ns = next_marker_start_after(all_starts, s)
            chunk = refs[s:ns] if ns is not None else refs[s:]
            chunk = chunk.strip()

            # apply your deterministic truncators (same as builder, but minimal)
            trunc, did_trunc, _ = truncate_at_embedded_marker_with_pos(chunk)
            if did_trunc:
                chunk = trunc
            chunk = truncate_at_and_followed_by_author_list(chunk)
            chunk = truncate_at_second_author_block(chunk)

            sc = _payload_score(chunk)

            if sc > best_score:
                best_score = sc
                best = (s, e)

        chosen[n] = best

    return chosen
```

```python
_AFFIL_TRIG = re.compile(
    r"(?i)\b("
    r"author information|corresponding author|"
    r"orcid|e-?mail|email|"
    r"department of|school of|college of|faculty of|"
    r"university|institute|laboratory|centre|center"
    r")\b"
)

_ACK_TRIG = re.compile(
    r"(?i)\b("
    r"acknowledg(?:ment|ments|e|es)|we thank|"
    r"supported by|funded by|funding|grant(s)?\b|"
    r"this work was supported|this research was supported"
    r")\b"
)

def has_min_citation_payload(prefix: str) -> bool:
    """
    Very cheap 'did we already capture the citation?' test.
    We only truncate if we have enough before the trigger.
    """
    p = prefix.strip()
    if len(p) < 60:
        return False
    # year OR journal-ish OR DOI-ish OR multiple author separators
    if re.search(r"\b(19|20)\d{2}\b", p):
        return True
    if re.search(r"\bdoi\b|10\.\d{4,9}/", p, re.I):
        return True
    if p.count(";") >= 3:
        return True
    if re.search(r"\bJ\.\b|\bProc\.\b|\bRev\.\b|\bChem\b|\bPolym\b|\bNature\b|\bScience\b", p, re.I):
        return True
    return False

def truncate_at_boilerplate_tail(s: str) -> str:
    """
    If affiliation/ack boilerplate appears, truncate the entry at that point,
    but ONLY if the prefix already looks like a citation.
    """
    if not s:
        return s

    # Find earliest trigger
    m1 = _AFFIL_TRIG.search(s)
    m2 = _ACK_TRIG.search(s)
    ms = [m for m in [m1, m2] if m is not None]
    if not ms:
        return s

    m = min(ms, key=lambda x: x.start())

    # Don't truncate if trigger is too close to start
    if m.start() < 30:
        return s

    prefix = s[:m.start()].rstrip()
    if not has_min_citation_payload(prefix):
        return s

    return prefix
```

```python
_DASH = re.compile(r"[−–-]")  # minus/en-dash/hyphen
_RESUME = re.compile(r"\b[A-Z][a-z]{2,}\b")  # candidate word to resume title/journal
AUTHOR_TOKEN = re.compile(
    r"\b[A-Z][A-Za-z'`-]+,\s*[A-Z](?:\.-?[A-Z]\.)?(?:\.)?\s*;?"
)

ACK_BLOCK = re.compile(
    r"(?is)\b("
    r"we thank|acknowledg(?:ment|ments|e|es)|supported by|funded by|funding|grant(s)?|"
    r"this work was supported|this research was supported|conflicts? of interest|"
    r"fee for funding|funding.*fee|"
    r"author information|corresponding author|orcid|e-?mail|email"
    r")\b"
)

_ZIP = re.compile(r"\b\d{4,6}\b")
_COUNTRY = re.compile(r"(?i)\b(united\b(?:\s+states\b)?|usa|uk|germany|france|china|japan|canada|australia|switzerland)\b")

def remove_inline_affiliation_insertion(s: str, max_span: int = 2500) -> str:
    """
    Remove inline '— Department/University/Institute...' blobs and resume
    AFTER a postal-code/country cue when present (more reliable than first capital word).
    """
    if not s:
        return s

    m = _AFFIL_TRIG.search(s)
    if not m:
        return s

    # dash just before trigger
    dash_pos = -1
    for ch in ["−", "–", "-"]:
        p = s.rfind(ch, 0, m.start())
        if p > dash_pos:
            dash_pos = p
    if dash_pos < 0:
        return s

    # scan forward from trigger for a better resume anchor
    scan = s[m.end(): m.end() + max_span]

    # Prefer: resume after ZIP or country token
    z = _ZIP.search(scan)
    c = _COUNTRY.search(scan)
    anchor = None
    if z and c:
        anchor = min(z.end(), c.end())
    elif z:
        anchor = z.end()
    elif c:
        anchor = c.end()

    if anchor is None:
        return s  # too risky to guess

    # after the anchor, resume at the next decent capitalized word
    tail2 = scan[anchor:]
    r = _RESUME.search(tail2)
    if not r:
        return s

    resume = m.end() + anchor + r.start()

    # require we actually removed a meaningful chunk
    if resume - dash_pos < 80:
        return s

    return (s[:dash_pos].rstrip() + " " + s[resume:].lstrip()).strip()


def excise_ack_affil_insertion(s: str, max_scan: int = 4000) -> str:
    """
    Remove an ACK insertion inside a reference and resume at the next valid author token.
    Prefer ACK triggers (we thank/funding/etc). Only fallback to affiliation triggers
    if no ACK trigger exists.
    """
    if not s:
        return s

    m = ACK_BLOCK.search(s)
    if m is None:
        m = _AFFIL_TRIG.search(s)
    if m is None:
        return s

    cut_start = m.start()
    if cut_start < 25:
        return s

    tail = s[m.end(): m.end() + max_scan]

    # Resume at FIRST valid author token after the ACK block
    mm = AUTHOR_TOKEN.search(tail)
    if not mm:
        return s

    resume_pos = m.end() + mm.start()
    return (s[:cut_start].rstrip() + " " + s[resume_pos:].lstrip()).strip()


def drop_leading_orphan_initials(s: str) -> str:
    """
    If the body starts with orphan initials like 'F.;' or 'A.;' (common from broken author lists),
    drop them.
    """
    if not s:
        return s
    m = re.match(r"^\s*(\(\s*\d+\s*\))\s*(.*)$", s)
    if not m:
        return s
    marker, body = m.group(1), m.group(2).lstrip()
    body2 = re.sub(r"^(?:[A-Z]\.\s*;?\s*){1,3}", "", body)  # drop 1–3 initial tokens
    return f"{marker} {body2}".strip()

```

```python
def build_entries_deterministic_bracketed(
    pdf_path: str,
    expected_n: int,
    method: str = "pdfplumber"
) -> dict[int, RefSpan]:

    raw = pdf_to_text(pdf_path, method=method)
    refs = extract_references_section(raw)
    refs = trim_to_first_reference(refs)
    refs = normalize_for_splitting(refs)

    marker_pos = select_markers_by_number_scored(refs, expected_n)  # {n: (start, end)}
    all_starts = build_all_marker_starts(refs, expected_n)          # sorted list of (start_idx, n)

    out: dict[int, RefSpan] = {}

    for n in range(1, expected_n + 1):
        notes: list[str] = []
        det_conf = 1.0

        p1 = marker_pos.get(n)

        # If marker missing, emit an empty span (do NOT emit a string)
        if p1 is None:
            out[n] = RefSpan(n=n, raw="", det_conf=0.0, notes=["marker_missing"])
            continue

        start_n, _end_n = p1

        next_start = next_marker_start_after(all_starts, start_n)

        chunk = refs[start_n:next_start] if next_start is not None else refs[start_n:]
        chunk2 = chunk.strip()

        # Boundary hygiene: if an embedded (k) appears inside this chunk, truncate there.
        trunc, did_trunc, cut_pos = truncate_at_embedded_marker_with_pos(chunk2)
        if did_trunc:
            notes.append("embedded_marker_truncation")
            # embedded marker truncation reduces confidence slightly
            det_conf = min(det_conf, 0.85)
            chunk2 = trunc

        # Very light normalization only (safe)
        chunk2 = strip_comma_after_marker(chunk2)
        chunk2 = clean_entry_det(chunk2)
        chunk2 = ensure_marker(n, chunk2)

        # Optional heuristic: if chunk is suspiciously short, lower confidence
        if len(chunk2) < 40:
            notes.append("suspiciously_short")
            det_conf = min(det_conf, 0.6)

        out[n] = RefSpan(n=n, raw=chunk2, det_conf=det_conf, notes=notes)

    return out



def extraction_report(entries_by_n: dict[int,str], expected_n: int) -> dict:
    missing = [n for n in range(1, expected_n + 1) if n not in entries_by_n]

    blank = []
    for n in range(1, expected_n + 1):
        t = entries_by_n.get(n, "")
        if isinstance(t, RefSpan):
            t = t.raw or ""
        body = re.sub(r"^\(\s*\d+\s*\)\s*", "", t).strip()

        if len(body) < 20:
            blank.append(n)

    return {
        "expected_n": expected_n,
        "present_n": len(entries_by_n),
        "missing_numbers": missing,
        "blank_numbers": blank,
    }
```

## LLM patch (Phi-3 via Ollama)

This stage is **only** used to patch references that are **missing** or **blank** after deterministic extraction.

**PASS criteria (builder stage):**
- Final `entries_final` contains **all** `1..EXPECTED_N`
- Each entry is non-empty and starts with the correct marker `(n)`

> Note: This stage is *not* the verification stage. It just ensures the DataFrame builder has a complete set of reference strings.


```python
import json
import time
import re
from typing import Optional

import pandas as pd
import requests

OLLAMA_URL = "http://localhost:11434"
OLLAMA_MODEL = "phi3"  # change if you use a different local model


def _ollama_chat(messages, model: str = OLLAMA_MODEL, temperature: float = 0.0, timeout: int = 120) -> str:
    """
    Call Ollama /api/chat (non-streaming). Requires local Ollama running.
    """
    url = f"{OLLAMA_URL}/api/chat"
    payload = {
        "model": model,
        "messages": messages,
        "stream": False,
        "options": {"temperature": temperature},
    }
    resp = requests.post(url, json=payload, timeout=timeout)
    resp.raise_for_status()
    data = resp.json()
    return (data.get("message", {}) or {}).get("content", "") or ""


def _context_window(refs_text: str, n: int, *, window: int = 4500) -> str:
    """
    Extract a local context window around (n). If (n) isn't found, use neighboring markers.
    """
    t = refs_text or ""
    if not t:
        return ""

    pat = re.compile(rf"\(\s*{n}\s*\)")
    m = pat.search(t)

    if m is None:
        # Try neighbors as anchors (closest first)
        for k in (n - 1, n + 1, n - 2, n + 2):
            if k < 1:
                continue
            mk = re.search(rf"\(\s*{k}\s*\)", t)
            if mk:
                m = mk
                break

    if m is None:
        # Fallback to last chunk (often where refs live)
        return t[-window:]

    start = max(0, m.start() - window // 2)
    end = min(len(t), m.start() + window // 2)
    return t[start:end]


# Single public name used everywhere else
def context_window_for_n(refs_text: str, n: int, window: int = 4500) -> str:
    return _context_window(refs_text, n, window=window)


def _normalize_for_match(s: str) -> str:
    return re.sub(r"\s+", "", s or "").lower()


def llm_extract_reference_n(
    *,
    n: int,
    refs_text: str,
    left: Optional[str] = None,
    right: Optional[str] = None,
    model: str = OLLAMA_MODEL
) -> tuple[str, dict]:
    """
    Use Phi-3 (Ollama) to extract EXACTLY reference (n) from a local refs_text window.
    Returns: (reference_string, meta_dict)
    """
    ctx = _context_window(refs_text, n, window=4500)

    system = (
        "You extract bibliographic references from messy PDF text. "
        "Use ONLY the provided text. Do NOT invent missing information. "
        "Return JSON only."
    )

    user = f"""
Task: Extract reference number ({n}) from the text below.

Rules:
- Use ONLY the provided text. If you cannot find reference ({n}), return an empty string.
- Output JSON with keys: n (int), reference (string), confidence (0..1), notes (string).
- The 'reference' must start with '({n})' exactly (no leading spaces).
- The reference must NOT contain any other reference marker like '(k)' for k != {n}.
- Remove obvious non-reference boilerplate (AUTHOR INFORMATION, ACKNOWLEDGMENTS, affiliations, Figure S/Table S callouts) IF they are clearly not part of the citation.
- Keep DOI/URL if present.

Neighbor anchors (may help):
LEFT (n-1): {left or ""}
RIGHT (n+1): {right or ""}

TEXT WINDOW:
<<<
{ctx}
>>>
""".strip()

    messages = [
        {"role": "system", "content": system},
        {"role": "user", "content": user},
    ]

    t0 = time.time()
    raw = _ollama_chat(messages, model=model, temperature=0.0, timeout=180)
    dt = time.time() - t0

    meta = {"n": n, "model": model, "elapsed_s": round(dt, 3), "raw_len": len(raw)}

    # Parse JSON
    try:
        j = json.loads(raw)
    except Exception:
        meta["parse_error"] = True
        meta["raw"] = raw[:500]
        return "", meta

    ref = (j.get("reference") or "").strip()
    conf = j.get("confidence", None)
    notes = j.get("notes", "")

    meta.update({"confidence": conf, "notes": notes})

    # Basic validation
    if not ref.startswith(f"({n})"):
        meta["invalid_marker"] = True
        return "", meta

    # Reject if it contains other markers
    others = [int(x) for x in re.findall(r"\(\s*(\d+)\s*\)", ref)]
    if any(k != n for k in others):
        meta["embedded_markers"] = others
        return "", meta

    # Light "evidence" check: does the first chunk appear in the context (ignoring whitespace)?
    probe = _normalize_for_match(ref[:120])
    ctxn = _normalize_for_match(ctx)
    if probe and probe[:40] and probe[:40] not in ctxn:
        meta["evidence_weak"] = True

    return ref, meta


def build_quality_dataframe(entries: dict[int, str], expected_n: int) -> pd.DataFrame:
    """
    Build a per-reference quality diagnostic DataFrame.
    Does NOT mutate entries.
    """
    def quality_flags(entry: str) -> dict:
        if isinstance(entry, RefSpan):
            t = entry.raw or ""
        else:
            t = entry or ""

        flags = {}

        flags["embedded_marker"] = bool(re.search(r"\(\s*\d+\s*\)", t[6:]))
        flags["has_author_info"] = bool(re.search(r"(?i)\bAUTHOR INFORMATION\b", t))
        flags["has_ack"] = bool(re.search(r"(?i)\bACKNOWLEDGMENTS\b|\bFunding\b", t))

        flags["captionish"] = bool(re.search(
                r"(?i)\b("
                r"photo|figure|scheme|table|supporting information|"
                r"contact angle|surface energy|peel force|displacement|"
                r"confocal|microscopy|image|images|"
                r"substrate|system"
                r")\b",
                t
        ))

        years = re.findall(r"\b(19|20)\d{2}\b", t)
        flags["multi_year"] = len(years) >= 2

        flags["too_short"] = len(t.strip()) < 120
        flags["empty_parens"] = "( )" in t or "()" in t

        flags["flagged"] = (
            flags["embedded_marker"]
            or flags["multi_year"]
            or flags["empty_parens"]
            or flags["captionish"]
            or flags["has_author_info"]
            or flags["has_ack"]
            or flags["too_short"]
        )

        return flags


    rows = []
    for n in range(1, expected_n + 1):
        span = entries.get(n, RefSpan(n=n, raw="", det_conf=0.0, notes=["missing_key"]))
        e = span.raw or ""
        f = quality_flags(span)
        rows.append({
            "n": n,
            "len": len(e),
            **f,
            "preview": e[:140],
            "det_conf": getattr(span, "det_conf", None),
            "notes": ";".join(getattr(span, "notes", []) or []),
        })


    return pd.DataFrame(rows)

```

```python
def phi3_prompt(n: int, window_text: str) -> str:
    return f"""
You are extracting ONE bibliography reference from messy PDF text.

Task:
- Extract ONLY reference number ({n}) from the text below.
- Use ONLY the provided text. Do NOT invent missing information.
- Return STRICT JSON only.

Output JSON:
{{
  "n": {n},
  "reference": "(%d) ...",
  "confidence": 0.0,
  "notes": ""
}}

Rules:
- reference MUST start with "({n})" (no leading spaces).
- reference must NOT contain any other "(k)" markers for k != {n}.
- Remove obvious non-reference boilerplate (AUTHOR INFORMATION, ACKNOWLEDGMENTS, affiliations, figure/table callouts) if clearly not part of the citation.
- Keep DOI/URL if present.
- If you cannot confidently extract ({n}) from the window, return reference="" and confidence<=0.4.

TEXT WINDOW:
<<<
{window_text}
>>>
""".strip()
```

```python
DOI_RE = re.compile(r"\b10\.\d{4,9}/[-._;()/:A-Z0-9]+\b", re.IGNORECASE)
URL_RE = re.compile(r"https?://\S+", re.IGNORECASE)

def needs_phi3(span: RefSpan) -> bool:
    raw = span.raw or ""
    # Always call if det confidence low
    if span.det_conf and span.det_conf < 0.6:
        return True
    # Very short chunks are suspicious
    if len(raw.strip()) < 40:
        return True
    # Multiple years or multiple DOIs often means glued refs
    years = re.findall(r"\b(19|20)\d{2}\b", raw)
    if len(years) >= 2:
        return True
    if len(DOI_RE.findall(raw)) >= 2:
        return True
    # If it contains a URL/DOI but it looks broken (common PDF line wrap issues)
    if "10." in raw and not DOI_RE.search(raw):
        return True
    if "http" in raw and not URL_RE.search(raw):
        return True
    return False
```

```python
def run_pipeline(pdf_path: str, expected_n: Optional[int] = None) -> Tuple[List[Dict[str, Any]], Dict[str, Any]]:
    spans: Dict[int, RefSpan] = build_entries_deterministic_bracketed(pdf_path, expected_n)

    parsed: List[Dict[str, Any]] = []
    phi3_used = 0

    for n in sorted(spans.keys()):
        span = spans[n]
        if needs_phi3(span):
            rec = phi3_parse_reference(span.raw, n)
            phi3_used += 1
        else:
            # Minimal “det-only” record (still structured)
            rec = {
                "n": n,
                "repaired": span.raw.strip(),
                "authors": None,
                "title": None,
                "journal": None,
                "year": None,
                "volume": None,
                "issue": None,
                "pages": None,
                "doi": None,
                "url": None,
                "confidence": "low" if span.det_conf < 0.8 else "medium",
                "notes": ["deterministic_only"]
            }
        parsed.append(rec)

    # Stage C validation (stub hook)
    # validated = run_verification_on_parsed(parsed)  # you can wire this to your DOI/URL checker

    report = {
        "expected_n": expected_n,
        "det_n": len(spans),
        "phi3_used": phi3_used,
        "phi3_rate": (phi3_used / max(1, len(spans))),
    }
    return parsed, report
```

```python
import json, re
from typing import Optional, Iterable, Tuple, Dict, Any

# ---------- helpers ----------
# def as_text(x) -> str:
#     """Return plain text from either RefSpan or string."""
#     if isinstance(x, RefSpan):
#         return x.raw or ""
#     return x or ""

def _safe_json_loads(s: str) -> Optional[dict]:
    """Try hard to parse JSON from model output (handles extra chatter)."""
    if not s:
        return None
    s = s.strip()

    # fast path
    try:
        return json.loads(s)
    except Exception:
        pass

    # try to extract the first {...} block
    m = re.search(r"\{.*\}", s, flags=re.DOTALL)
    if not m:
        return None
    blob = m.group(0)

    # remove trailing junk after last }
    try:
        return json.loads(blob)
    except Exception:
        return None

DOI_RE = re.compile(r"\b10\.\d{4,9}/[-._;()/:A-Z0-9]+\b", re.IGNORECASE)
URL_RE = re.compile(r"https?://\S+", re.IGNORECASE)
YEAR_RE = re.compile(r"\b(19|20)\d{2}\b")

def _normalize_doi(doi: Optional[str]) -> Optional[str]:
    if not doi:
        return None
    doi = doi.strip()
    doi = re.sub(r"^(https?://(dx\.)?doi\.org/)", "", doi, flags=re.IGNORECASE)
    m = DOI_RE.search(doi)
    return m.group(0) if m else doi

def _normalize_url(url: Optional[str]) -> Optional[str]:
    if not url:
        return None
    return url.strip().rstrip(").,;")

# ---------- canonical phi3 parser ----------
def phi3_parse_reference(
    raw_ref: str,
    n: int,
    left: str = "",
    right: str = "",
    model: str = None,
) -> Dict[str, Any]:
    """
    Phi3 parses/repairs a single reference chunk.
    Contract: do NOT invent; output strict JSON; nulls allowed.
    """
    raw_ref = (raw_ref or "").strip()
    left = (left or "").strip()
    right = (right or "").strip()

    prompt = f"""
You are cleaning and parsing ONE bibliography reference.

CRITICAL EVIDENCE RULE:
- For every field you fill (authors/title/journal/year/volume/issue/pages/doi/url),
  the exact supporting substring MUST appear in TARGET reference text.
- If it does not appear verbatim in TARGET, set the field to null.
- Do NOT use left/right to populate fields; left/right are only for boundary detection.

HARD RULES:
- DO NOT invent missing authors/title/journal/year/volume/pages/doi/url.
- Use only what is explicitly present in the provided text.
- If unsure/missing, set the field to null.
- Return STRICT JSON ONLY. No markdown. No commentary.

REPAIR RULES:
- You may DELETE non-bibliographic phrases (figure captions, SI labels, instrument notes).
- You may NOT ADD any bibliographic content not present in TARGET.

Reference number: {n}

Left neighbor (may help detect glue / boundaries):
{left}

TARGET reference text (primary):
{raw_ref}

Right neighbor (may help detect glue / boundaries):
{right}

Return JSON with exactly these keys:
{{
  "n": {n},
  "repaired": "string or null",
  "authors": "string or null",
  "title": "string or null",
  "journal": "string or null",
  "year": "int or null",
  "volume": "string or null",
  "issue": "string or null",
  "pages": "string or null",
  "doi": "string or null",
  "url": "string or null",
  "confidence": "low|medium|high",
  "notes": ["..."]
}}
""".strip()

    # transport
    msgs = [{"role": "user", "content": prompt}]

    if "_ollama_chat" in globals():
        out = _ollama_chat(msgs, model=model) if model else _ollama_chat(msgs)
    else:
        out = ollama_chat(msgs, model=model) if model else ollama_chat(msgs)

    rec = _safe_json_loads(out)
    if not rec:
        # fallback record; keep raw text but flag failure
        return {
            "n": n,
            "repaired": raw_ref if raw_ref else None,
            "authors": None, "title": None, "journal": None, "year": None,
            "volume": None, "issue": None, "pages": None,
            "doi": None, "url": None,
            "confidence": "low",
            "notes": ["phi3_json_parse_failed"],
            "_raw_model_output": out[:800] if isinstance(out, str) else str(out)[:800],
        }

    # normalize fields
    rec["n"] = n
    if "doi" in rec:
        rec["doi"] = _normalize_doi(rec.get("doi"))
    if "url" in rec:
        rec["url"] = _normalize_url(rec.get("url"))

    raw_lower = raw_ref.lower()

    doi = rec.get("doi")
    if doi:
        # accept only if the DOI (or its core) appears in raw_ref
        if doi.lower() not in raw_lower and f"doi.org/{doi.lower()}" not in raw_lower:
            rec.setdefault("notes", []).append("doi_not_in_raw_ref_nulling")
            rec["doi"] = None

    url = rec.get("url")
    if url:
        if url.lower() not in raw_lower:
            rec.setdefault("notes", []).append("url_not_in_raw_ref_nulling")
            rec["url"] = None


    # ensure repaired exists
    repaired = (rec.get("repaired") or "").strip()
    if not repaired:
        # If model didn't provide repaired, default to raw
        rec["repaired"] = raw_ref if raw_ref else None
        repaired = (rec.get("repaired") or "").strip()

    # cheap sanity: if repaired contains multiple DOIs/years, flag it
    if repaired:
        if len(DOI_RE.findall(repaired)) >= 2:
            rec.setdefault("notes", []).append("possible_glued_refs_multiple_doi")
            rec["confidence"] = "low"
        if len(YEAR_RE.findall(repaired)) >= 2:
            rec.setdefault("notes", []).append("possible_glued_refs_multiple_year")
            rec["confidence"] = "low"

    return rec

# ---------- canonical patcher ----------
def patch_with_phi3(
    entries_det: dict[int, RefSpan],
    expected_n: int,
    refs_text: str,  # kept for signature compatibility; not used here unless you add windowing
    targets: Optional[Iterable[int]] = None,
    min_len: int = 120,
    model: str = None,
) -> Tuple[dict[int, str], dict[str, Any], dict[int, Dict[str, Any]]]:
    """
    Patch selected references by calling phi3_parse_reference on bounded spans.
    Returns:
      entries_final: dict[int, str]
      report: dict
      parsed_records: dict[int, dict]  (structured output per n)
    """

    entries_final: dict[int, str] = {}
    parsed_records: dict[int, Dict[str, Any]] = {}

    # initialize from determinism
    for n in range(1, expected_n + 1):
        entries_final[n] = as_text(entries_det.get(n, RefSpan(n=n, raw=""))).strip()

    patch_list = sorted(set(targets or []))

    # include blanks always
    for n in range(1, expected_n + 1):
        if not entries_final[n]:
            patch_list.append(n)
    patch_list = sorted(set(patch_list))

    filled, failed, skipped = [], [], []

    for n in patch_list:
        current = entries_final.get(n, "")
        forced = (targets is not None and n in set(targets))

        if (not forced) and current and len(current) >= min_len:
            skipped.append(n)
            continue


        raw_in = as_text(entries_det.get(n, RefSpan(n=n, raw=""))).strip()
        if not raw_in:
            failed.append(n)
            parsed_records[n] = {"n": n, "confidence": "low", "notes": ["no_raw_span_to_patch"]}
            continue

        left = as_text(entries_final.get(n - 1, ""))[:300] if n > 1 else ""
        right = as_text(entries_final.get(n + 1, ""))[:300] if n < expected_n else ""

        rec = phi3_parse_reference(raw_in, n=n, left=left, right=right, model=model)
        parsed_records[n] = rec

        repaired = (rec.get("repaired") or "").strip()
        if repaired and len(repaired) >= 40:
            entries_final[n] = repaired
            filled.append(n)
        else:
            failed.append(n)

    final_report = extraction_report(entries_final, expected_n)

    report = {
        "patch_list": patch_list,
        "filled": filled,
        "failed": failed,
        "skipped": skipped,
        "final_report": final_report,
    }
    return entries_final, report, parsed_records
```

```python
import re
import json
import pandas as pd
from typing import Optional, Any, Dict, List
from urllib.parse import urlparse

# -------------------------
# Deterministic extractors
# -------------------------

DOI_RE  = re.compile(r"\b10\.\d{4,9}/[^\s\"<>]+", re.IGNORECASE)
URL_RE  = re.compile(r"(https?://[^\s\]\)>,;]+)", re.IGNORECASE)
YEAR_RE = re.compile(r"\b(19|20)\d{2}\b")

def _det_year(text: str) -> Optional[int]:
    if not text:
        return None
    ys = [int(m.group(0)) for m in YEAR_RE.finditer(text)]
    return ys[0] if ys else None

def _det_doi(text: str) -> Optional[str]:
    if not text:
        return None
    m = DOI_RE.search(text)
    if not m:
        return None
    doi = m.group(0).strip().rstrip(").,;")
    doi = re.sub(r"^(https?://(dx\.)?doi\.org/)", "", doi, flags=re.IGNORECASE)
    return doi or None

def _det_urls(text: str) -> List[str]:
    if not text:
        return []
    urls = [m.group(1).strip().rstrip(").,;") for m in URL_RE.finditer(text)]
    out, seen = [], set()
    for u in urls:
        if u not in seen:
            out.append(u)
            seen.add(u)
    return out

def _det_authors_guess(text: str) -> Optional[str]:
    if not text:
        return None
    t = re.sub(r"^\(\s*\d+\s*\)\s*", "", text).strip()
    cut = len(t)

    ym = YEAR_RE.search(t)
    if ym: cut = min(cut, ym.start())
    um = URL_RE.search(t)
    if um: cut = min(cut, um.start())
    dm = DOI_RE.search(t)
    if dm: cut = min(cut, dm.start())

    guess = t[:cut].strip(" .;,:-")
    return guess or None

def _det_title_guess(text: str) -> Optional[str]:
    if not text:
        return None
    t = re.sub(r"^\(\s*\d+\s*\)\s*", "", text).strip()
    t = URL_RE.sub("", t).strip()

    parts = [p.strip() for p in re.split(r"\.\s+", t) if p.strip()]
    if len(parts) < 2:
        return None

    # choose a likely title-ish chunk
    for c in sorted(parts, key=len, reverse=True):
        if c.count(";") >= 4:
            continue
        if len(c) < 12:
            continue
        return c.strip(" .;,:-") or None
    return None

def _det_journal_guess(text: str) -> Optional[str]:
    if not text:
        return None
    urls = _det_urls(text)
    if not urls:
        return None
    try:
        host = urlparse(urls[0]).netloc.lower().replace("www.", "")
        return host or None
    except Exception:
        return None


# -------------------------
# Small helpers
# -------------------------

def _as_str(x: Any) -> Optional[str]:
    if x is None:
        return None
    s = str(x).strip()
    return s if s else None

def _as_int(x: Any) -> Optional[int]:
    if x is None:
        return None
    if isinstance(x, int):
        return x
    s = str(x).strip()
    if not s:
        return None
    m = re.search(r"(19|20)\d{2}", s)
    return int(m.group(0)) if m else None

def _as_list_of_str(x: Any) -> List[str]:
    if x is None:
        return []
    if isinstance(x, list):
        return [str(v).strip() for v in x if str(v).strip()]
    s = str(x).strip()
    return [s] if s else []

def _jsonable_list(x: Any) -> List[str]:
    # Ensures parquet/pyarrow friendliness
    xs = _as_list_of_str(x)
    out, seen = [], set()
    for v in xs:
        if v not in seen:
            out.append(v)
            seen.add(v)
    return out

def _clean_authors(authors: Optional[str], raw_ref: Optional[str] = None) -> Optional[str]:
    """
    Return an author-only string, trimming anything that looks like it belongs to the
    title/journal/year/doi/url/pages portion of the reference.

    Strategy:
      1) If raw_ref is available, take the prefix of raw_ref up to the first year/doi/url.
      2) Otherwise, trim authors string at the first strong citation delimiter.
      3) Normalize separators.
    """
    if not authors and not raw_ref:
        return None

    # Prefer using the raw reference when available (most reliable)
    if raw_ref:
        s = re.sub(r"^\(\s*\d+\s*\)\s*", "", raw_ref).strip()

        # find earliest boundary indicating we're past authors
        cuts = []

        m = YEAR_RE.search(s)
        if m:
            cuts.append(m.start())

        m = DOI_RE.search(s)
        if m:
            cuts.append(m.start())

        m = URL_RE.search(s)
        if m:
            cuts.append(m.start())

        # common patterns indicating journal/volume/pages region even without year
        m = re.search(r"\b\d+\s*,\s*\d+\b", s)  # "18, 1234"
        if m:
            cuts.append(m.start())

        m = re.search(r"\b\d+\s*[-–]\s*\d+\b", s)  # pages like "123-130"
        if m:
            cuts.append(m.start())

        if cuts:
            s = s[:min(cuts)].strip(" .;,:-")

        # If we ended up with something reasonable, treat it as authors region
        if s:
            authors = s

    if not authors:
        return None

    s = str(authors).strip()

    # If title leaked in, it often introduces longer words with spaces after a run of initials.
    # Conservative cut: stop at the first occurrence of two consecutive spaces-separated words
    # that aren't initial-like (e.g., "Synthesis of", "Extremely Softened", etc.)
    m = re.search(r";\s*[A-Z][a-z]{3,}\s+[A-Z][a-z]{3,}", s)
    if m:
        s = s[:m.start()].strip(" .;,:-")

    # Normalize delimiters/spaces
    s = re.sub(r"\s*;\s*", "; ", s)
    s = re.sub(r"\s*,\s*", ", ", s)
    s = re.sub(r"\s+", " ", s).strip(" .;,:-")

    return s or None



def _fallback_title_from_raw(raw: str) -> Optional[str]:
    """Last-resort query string so scrapeNverify isn't stuck with no title/doi/url."""
    if not raw:
        return None

    # remove "(n)" prefix
    t = re.sub(r"^\(\s*\d+\s*\)\s*", "", raw).strip()

    # drop urls / doi
    t = URL_RE.sub("", t)
    t = DOI_RE.sub("", t)

    # normalize whitespace
    t = re.sub(r"\s+", " ", t).strip(" .;,:-")

    if len(t) < 20:
        return None

    return t[:220]



# -------------------------
# Final DataFrame builder
# -------------------------

def build_refs_dataframe_final(
    entries_det: Dict[int, Any],                    # dict[int, RefSpan]
    expected_n: int,
    entries_final: Optional[Dict[int, str]] = None, # dict[int, str]
    phi3_records: Optional[Dict[int, Dict[str, Any]]] = None,  # dict[int, dict]
) -> pd.DataFrame:
    """
    Build the analysis-ready DataFrame with explicit separation:
      - deterministic fields (*_det / *_guess_det)
      - LLM-derived fields (*_llm)
      - conservative finals (*_final)

    Parquet-safe: lists are used for urls/notes (pyarrow ok). If your environment
    lacks pyarrow list support, switch lists to JSON strings in one place.
    """
    rows: List[Dict[str, Any]] = []

    entries_final = entries_final or {}
    phi3_records = phi3_records or {}

    for ref_id in range(1, expected_n + 1):
        span = entries_det.get(ref_id)

        raw_det = (getattr(span, "raw", "") or "").strip() if span is not None else ""
        raw_patched = (entries_final.get(ref_id, "") or "").strip()

        if raw_patched and _looks_like_ref_for_id(raw_patched, ref_id):
            raw_final = raw_patched
        else:
            raw_final = raw_det


        det_conf   = getattr(span, "det_conf", None) if span is not None else None
        det_notes  = getattr(span, "notes", []) if span is not None else []
        start_idx  = getattr(span, "start_idx", None) if span is not None else None
        end_idx    = getattr(span, "end_idx", None) if span is not None else None

        # Deterministic extraction should run on the final raw text (better if patched),
        # but we keep the outputs as *_det to indicate provenance.
        year_det          = _det_year(raw_final)
        doi_det           = _det_doi(raw_final)
        urls_det          = _det_urls(raw_final)
        authors_guess_det = _det_authors_guess(raw_final)
        title_guess_det   = _det_title_guess(raw_final)
        journal_guess_det = _det_journal_guess(raw_final)

        # LLM record (only present if patched)
        rec = phi3_records.get(ref_id) or {}

        repaired_llm = _as_str(rec.get("repaired"))
        authors_llm  = _as_str(rec.get("authors"))
        title_llm    = _as_str(rec.get("title"))
        journal_llm  = _as_str(rec.get("journal"))
        year_llm     = _as_int(rec.get("year"))
        volume_llm   = _as_str(rec.get("volume"))
        issue_llm    = _as_str(rec.get("issue"))
        pages_llm    = _as_str(rec.get("pages"))
        doi_llm      = _as_str(rec.get("doi"))
        url_llm      = _as_str(rec.get("url"))

        # Conservative finals:
        # - DOI/URLs: deterministic only (LLM candidates kept separately)
        # - Year: use deterministic first, else LLM year (often explicitly present)
        # - Title: prefer LLM title if present (but still conservative: it’s just a parse)
        # - Authors/Journal: same pattern; finals are “best-available parse,” not verified truth
        year_final    = year_det if year_det is not None else year_llm
        title_final = title_llm or title_guess_det or _fallback_title_from_raw(raw_final)
        authors_final = _clean_authors(authors_llm or authors_guess_det, raw_ref=raw_final)
        journal_final = journal_llm if journal_llm is not None else journal_guess_det

        used_phi3 = bool(rec)

        rows.append({
            # identity
            "ref_id": ref_id,

            # provenance/raw
            "raw_det": raw_det,
            "raw_final": raw_final,
            "det_conf": float(det_conf) if det_conf is not None else None,
            "det_notes": _jsonable_list(det_notes),
            "start_idx": start_idx,
            "end_idx": end_idx,

            # deterministic (explicit)
            "year_det": year_det,
            "doi_det": doi_det,
            "urls_det": _jsonable_list(urls_det),
            "authors_guess_det": authors_guess_det,
            "title_guess_det": title_guess_det,
            "journal_guess_det": journal_guess_det,

            # llm (explicit)
            "used_phi3": used_phi3,
            "repaired_llm": repaired_llm,
            "authors_llm": authors_llm,
            "title_llm": title_llm,
            "journal_llm": journal_llm,
            "year_llm": year_llm,
            "volume_llm": volume_llm,
            "issue_llm": issue_llm,
            "pages_llm": pages_llm,
            "doi_llm": doi_llm,
            "url_llm": url_llm,

            # finals (conservative; not “verified truth”)
            "year_final": year_final,
            "doi_final": doi_det,                # IMPORTANT: deterministic only
            "urls_final": _jsonable_list(urls_det),  # IMPORTANT: deterministic only
            "authors_final": authors_final,
            "title_final": title_final,
            "journal_final": journal_final,

            # basic flags
            "is_missing": bool(not raw_final),
        })

    df = pd.DataFrame(rows)

    # stable ordering (nice for diffs)
    col_order = [
        "ref_id",
        "raw_det", "raw_final",
        "det_conf", "det_notes", "start_idx", "end_idx",
        "year_det", "doi_det", "urls_det", "authors_guess_det", "title_guess_det", "journal_guess_det",
        "used_phi3", "repaired_llm", "authors_llm", "title_llm", "journal_llm", "year_llm",
        "volume_llm", "issue_llm", "pages_llm", "doi_llm", "url_llm",
        "year_final", "doi_final", "urls_final", "authors_final", "title_final", "journal_final",
        "is_missing",
    ]
    df = df[[c for c in col_order if c in df.columns]]

    # ---- compatibility columns for downstream notebooks expecting legacy names ----
    # Keep scrapeNverify.ipynb unchanged by providing these aliases.
    df["raw"]     = df["raw_final"]
    df["authors"] = df["authors_final"]
    df["year"]    = df["year_final"]
    df["doi"]     = df["doi_final"]
    df["urls"]    = df["urls_final"]


    return df
```

```python
def _looks_like_ref_for_id(s: str, ref_id: int) -> bool:
    if not s:
        return False
    s = s.strip()

    # Strong signal: preserves "(n)" prefix
    if re.match(rf"^\(\s*{ref_id}\s*\)", s):
        return True

    # Fallback: author-like pattern "Lastname, X."
    return bool(re.match(r"^[A-Z][A-Za-z\-\']+,\s*[A-Z]", s))


df = build_refs_dataframe_final(
    entries_det=entries_det,
    expected_n=EXPECTED_N,
    entries_final=entries_final,
    phi3_records=phi3_records,
)

df.head(10)
```

```python
def debug_ref_pipeline_for_n(pdf_path: str, expected_n: int, n: int, method: str="pdfplumber"):
    raw = pdf_to_text(pdf_path, method=method)
    refs = normalize_for_splitting(trim_to_first_reference(extract_references_section(raw)))

    marker_pos = select_markers_by_number(refs, expected_n)
    all_starts = build_all_marker_starts(refs, expected_n)

    s, e = marker_pos[n]
    ns = next_marker_start_after(all_starts, s)
    chunk = refs[s:ns] if ns is not None else refs[s:]
    x = chunk.strip()

    def show(tag, val):
        print(f"\n--- {tag} --- len={len(val)}")
        print(val[:600])

    show("RAW SLICE", x)

    trunc, did_trunc, cut_pos = truncate_at_embedded_marker_with_pos(x)
    if did_trunc:
        x = trunc
    show("after embedded-marker trunc", x)

    x = truncate_at_and_followed_by_author_list(x)
    show("after truncate_at_and_followed_by_author_list", x)

    x = truncate_at_second_author_block(x)
    show("after truncate_at_second_author_block", x)

    x = remove_inline_affiliation_insertion(x)
    show("after remove_inline_affiliation_insertion", x)

    x = excise_ack_affil_insertion(x)
    show("after excise_ack_affil_insertion", x)

    x = drop_leading_orphan_initials(x)
    show("after drop_leading_orphan_initials", x)

    x = truncate_at_boilerplate_tail(x)
    show("after truncate_at_boilerplate_tail", x)

    x = drop_leading_title_fragment(x)
    show("after drop_leading_title_fragment", x)

    x = strip_comma_after_marker(x)
    x = clean_entry_det(x)
    x = ensure_marker(n, x)
    show("FINAL", x)
```
