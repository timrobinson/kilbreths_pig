# PDF bibliography → DataFrame (deterministic + LLM patch scaffold)
This notebook extracts numbered references from a PDF (ACS-style `(n)`), builds a DataFrame compatible with your Word pipeline, and optionally uses an LLM to patch missing/blank references.


```
import re
import pandas as pd
import pdfplumber

try:
    import fitz  # PyMuPDF
except Exception:
    fitz = None
```

```
def pdf_to_text(pdf_path: str, method: str = "pdfplumber") -> str:
    """Extract text from a text-native PDF."""
    if method == "pdfplumber":
        chunks = []
        with pdfplumber.open(pdf_path) as pdf:
            for page in pdf.pages:
                txt = page.extract_text() or ""
                txt = re.sub(r"\r\n?", "\n", txt)
                chunks.append(txt)
        return "\n\n".join(chunks)

    if method == "pymupdf":
        if fitz is None:
            raise RuntimeError("PyMuPDF not available")
        doc = fitz.open(pdf_path)
        return "\n\n".join(page.get_text("text") or "" for page in doc)

    raise ValueError(f"Unknown method: {method}")
```

```
def extract_references_section(text: str) -> str:
    """Return text starting at the first 'REFERENCES' occurrence."""
    t = text.replace("\r\n", "\n").replace("\r", "\n")
    m = re.search(r"(?mi)^\s*references\b", t)
    if m:
        return t[m.start():].strip()
    m = re.search(r"(?i)\breferences\b", t)
    if m:
        return t[m.start():].strip()
    raise ValueError("Could not find 'references' in extracted text.")
```

```
def trim_to_first_reference(refs_text: str) -> str:
    """Trim everything before the first *true* reference start."""
    t = refs_text.replace("\r\n", "\n").replace("\r", "\n")
    m = re.search(r"\(\s*1\s*\)\s*[A-Z][A-Za-z'’\-]+\s*,", t)
    if m:
        return t[m.start():].strip()
    m = re.search(r"\(\s*1\s*\)", t)
    if m:
        return t[m.start():].strip()
    raise ValueError("Couldn't find a '(1)' marker in references text.")
```

```
def normalize_for_splitting(text: str) -> str:
    """Keep newlines; only safe repairs."""
    t = text.replace("\r\n", "\n").replace("\r", "\n")
    t = re.sub(r"(\w)-\n(\w)", r"\1\2", t)   # join hyphenated line breaks
    t = re.sub(r"[ \t]+", " ", t)               # collapse spaces/tabs
    return t.strip()
```

```
def split_references_acs_safe(text: str) -> list[str]:
    """Split on '(n)' markers likely to be true reference starts; reject 53(19) issue numbers."""
    t = text.replace("\r\n", "\n").replace("\r", "\n").strip()

    matches = list(re.finditer(r"\(\s*\d+\s*\)", t))
    if not matches:
        raise ValueError("No '(n)' markers found.")

    starts = []
    for m in matches:
        i = m.start()
        prev = t[i-1] if i > 0 else ""
        if prev.isdigit():
            continue
        if i == 0 or prev.isspace() or prev in ".;:,":
            starts.append(i)

    if not starts:
        raise ValueError("No safe reference-start markers found.")

    out = []
    for idx, s in enumerate(starts):
        e = starts[idx + 1] if idx + 1 < len(starts) else len(t)
        out.append(t[s:e].strip())
    return out
```

```
def normalize_entry(entry: str) -> str:
    """Flatten newlines within an entry."""
    t = entry.replace("\r\n", "\n").replace("\r", "\n")
    t = re.sub(r"\s*\n\s*", " ", t)
    t = re.sub(r"[ \t]+", " ", t)
    return t.strip()
```

```
def ref_number(entry: str):
    m = re.match(r"^\(\s*(\d+)\s*\)", entry.strip())
    return int(m.group(1)) if m else None

def dedupe_keep_longest(entries: list[str]) -> list[str]:
    """If we over-split, keep the longest chunk for each ref number."""
    buckets = {}
    for e in entries:
        n = ref_number(e)
        if n is None:
            continue
        buckets.setdefault(n, []).append(e)
    best = {n: max(lst, key=len) for n, lst in buckets.items()}
    return [best[n] for n in sorted(best)]
```

```
def strip_embedded_other_markers(entry: str) -> str:
    """If another '(k)' appears later in the same chunk, cut before it."""
    n = ref_number(entry)
    if n is None:
        return entry.strip()
    for m in re.finditer(r"\(\s*(\d+)\s*\)", entry):
        k = int(m.group(1))
        if k != n and m.start() > 0:
            return entry[:m.start()].strip()
    return entry.strip()
```

```
_TRUNC_HEADINGS = re.compile(
    r"(?i)\b(AUTHOR INFORMATION|ACKNOWLEDGMENTS|Funding|Notes|Complete contact information is available)\b"
)

_SPAN_PATTERNS = [
    r"\bSupporting Information is available.*?(?=(\(\s*\d+\s*\))|$)",
    r"\bAdditional experimental details.*?(?=(\(\s*\d+\s*\))|$)",
    r"\bFigure\s*S\d+\b",
    r"\bFigures\s*S\d+(?:[–-]S\d+)?\b",
    r"\bTable\s*S\d+\b",
    r"\bTables\s*S\d+(?:\s*,\s*S\d+)*\b",
    r"\(PDF\)",
    r"\borcid\.org/\S+",
    r"\bEmail:\s*\S+",
]
_span_re = re.compile("|".join(_SPAN_PATTERNS), re.IGNORECASE)

def clean_entry_det(entry: str) -> str:
    """Deterministic cleanup. IMPORTANT: should not erase the entire ref."""
    t = entry.replace("\r\n", "\n").replace("\r", "\n")

    m = _TRUNC_HEADINGS.search(t)
    if m:
        t = t[:m.start()]

    t = _span_re.sub(" ", t)
    t = normalize_entry(t)
    return t
```

```
def ensure_marker(n: int, s: str) -> str:
    s = s.strip()
    if re.match(r"^\(\s*\d+\s*\)", s):
        return s
    return f"({n}) {s}"
```

```
def build_entries_deterministic(pdf_path: str, method: str = "pdfplumber") -> dict[int, str]:
    raw = pdf_to_text(pdf_path, method=method)
    refs = extract_references_section(raw)
    refs = trim_to_first_reference(refs)
    refs = normalize_for_splitting(refs)

    entries = split_references_acs_safe(refs)
    entries = [normalize_entry(e) for e in entries]
    entries = dedupe_keep_longest(entries)

    out = {}
    for e in entries:
        n = ref_number(e)
        if n is None:
            continue
        e = strip_embedded_other_markers(e)
        e = clean_entry_det(e)
        e = ensure_marker(n, e)
        out[n] = e
    return out

def extraction_report(entries_by_n: dict[int,str], expected_n: int) -> dict:
    present = sorted(entries_by_n.keys())
    missing = [n for n in range(1, expected_n+1) if n not in entries_by_n]
    blank = [n for n in present if len(re.sub(r"^\(\s*\d+\s*\)\s*", "", entries_by_n[n]).strip()) < 20]
    return {
        "expected_n": expected_n,
        "present_n": len(present),
        "missing_numbers": missing,
        "blank_numbers": blank,
    }
```

## Run deterministic extraction


```
PDF_PATH = "test.pdf"          # <-- change as needed
EXPECTED_N = 55                # <-- user-supplied

entries_det = build_entries_deterministic(PDF_PATH)
report = extraction_report(entries_det, EXPECTED_N)

report
```

```
def patch_with_llm(entries_by_n: dict[int,str], expected_n: int, refs_text: str) -> tuple[dict[int,str], dict]:
    report = extraction_report(entries_by_n, expected_n)
    to_patch = sorted(set(report["missing_numbers"] + report["blank_numbers"]))

    patched = dict(entries_by_n)
    patched_report = {"patched_numbers": [], "errors": {}}

    for n in to_patch:
        left = patched.get(n-1)
        right = patched.get(n+1)
        try:
            s = llm_extract_reference_n(n=n, refs_text=refs_text, left=left, right=right)
            s = normalize_entry(s)
            s = ensure_marker(n, s)
            patched[n] = s
            patched_report["patched_numbers"].append(n)
        except Exception as e:
            patched_report["errors"][n] = repr(e)

    final_report = extraction_report(patched, expected_n)
    return patched, {"initial": report, "patch": patched_report, "final": final_report}
```
