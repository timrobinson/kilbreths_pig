```python
import re
import json
import requests
import pandas as pd
from docx import Document
from typing import List, Dict, Optional, Tuple, Any
import pyarrow
from pathlib import Path
```

```python
DOI_RE = re.compile(r"\b(10\.\d{4,9}/[^\s\"<>]+)\b", re.IGNORECASE)
URL_RE = re.compile(r"\bhttps?://[^\s\)\]\}<>\"']+\b", re.IGNORECASE)
YEAR_RE = re.compile(r"\b(19|20)\d{2}\b")
DOI_TOKEN_RE = re.compile(r"\bdoi\s*:\s*(10\.\d{4,9}/\S+)",re.IGNORECASE)

REF_START_PATTERNS = [
    re.compile(r"^\s*\[\s*(\d+)\s*\]"),  # [1], [ 12 ]
    re.compile(r"^\s*(\d+)\s*\.\s+"),    # 1. ...
]

JOURNAL_YEAR_RE = re.compile(r"\.\s*([A-Z][^.;]{2,120}?)\.\s*(19|20)\d{2}\b")
```

```python
# ----------------------------
# 1) DOCX reading utilities
# ----------------------------

def load_docx_paragraphs(docx_path: str) -> List[Tuple[str, str]]:
    """
    Returns list of (style_name, text) for each paragraph in reading order.
    Keeps paragraph boundaries (more reliable than raw text extraction).
    """
    doc = Document(docx_path)
    out = []
    for p in doc.paragraphs:
        txt = (p.text or "").strip()
        style = getattr(p.style, "name", "") or ""
        out.append((style, txt))
    return out

def find_bibliography_block(
    paragraphs: List[Tuple[str, str]],
    heading_keywords: Tuple[str, ...] = ("bibliography", "references", "literature cited", "works cited"),
) -> List[str]:
    """
    Finds bibliography section even when the heading is styled as Normal and numbered,
    e.g. '6.0\\tBibliography'. Collects subsequent paragraphs until the next numbered
    section heading like '7.0 Something' (or end of doc).
    """

    def normalize(s: str) -> str:
        return re.sub(r"\s+", " ", (s or "").strip()).lower()

    # heading match: keyword anywhere in the normalized line
    def is_biblio_heading(txt: str) -> bool:
        t = normalize(txt)
        return any(k in t for k in heading_keywords)

    # stop when next section heading begins (e.g., 7.0 Conclusion, 7 Conclusion, 7.1 ...)
    next_section_re = re.compile(r"^\s*\d+(\.\d+)*\s+\S+")

    start_idx = None
    for i, (_, txt) in enumerate(paragraphs):
        if txt and is_biblio_heading(txt):
            start_idx = i
            break

    if start_idx is None:
        return []

    block = []
    for j in range(start_idx + 1, len(paragraphs)):
        _, txt = paragraphs[j]
        if not txt:
            continue
        # stop at next numbered heading (short line heuristic)
        if next_section_re.match(txt) and len(txt.split()) <= 10:
            break
        block.append(txt)

    return block


# ----------------------------
# 2) Reference splitting
# ----------------------------

def is_ref_start(line: str) -> Optional[int]:
    for pat in REF_START_PATTERNS:
        m = pat.search(line)
        if m:
            return int(m.group(1))
    return None


def split_references_from_block(block_paragraphs: List[str]) -> List[Dict]:
    """
    Build references by accumulating paragraphs until the next ref-start is encountered.
    Returns list of dicts: {ref_id, raw}
    """
    refs = []
    cur_lines = []
    cur_id = None

    for para in block_paragraphs:
        if not para:
            continue

        new_id = is_ref_start(para)
        if new_id is not None:
            # flush existing
            if cur_lines:
                refs.append({
                    "ref_id": cur_id,
                    "raw": " ".join(cur_lines).strip()
                })
            cur_lines = [para.strip()]
            cur_id = new_id
        else:
            # continuation line of current reference
            if cur_lines:
                cur_lines.append(para.strip())
            else:
                # sometimes bibliography begins without numbering; keep as "unkeyed"
                cur_lines = [para.strip()]
                cur_id = None

    if cur_lines:
        refs.append({"ref_id": cur_id, "raw": " ".join(cur_lines).strip()})

    # filter out junk/short lines
    refs = [r for r in refs if len(r["raw"]) > 10]
    return refs


# ----------------------------
# 3) Deterministic field extraction (no LLM)
# ----------------------------

def extract_doi(raw: str) -> Optional[str]:
    m = DOI_RE.search(raw)
    return m.group(0).rstrip(".,;") if m else None

def extract_urls(raw: str) -> List[str]:
    urls = [u.rstrip(".,;") for u in URL_RE.findall(raw)]
    # de-dup preserve order
    seen = set()
    out = []
    for u in urls:
        if u not in seen:
            seen.add(u)
            out.append(u)
    return out

def extract_year(raw: str) -> int | None:
    if not raw:
        return None
    years = [int(m.group(0)) for m in YEAR_RE.finditer(raw)]
    return max(years) if years else None


def heuristic_title(raw: str) -> Optional[str]:
    """
    Extract a plausible title from common reference formats.
    Avoid returning DOI/URL fragments as the title.
    """
    s = " ".join((raw or "").split())
    s = re.sub(r"^\[\s*\d+\s*\]\s*", "", s).strip()

    if not s:
        return None

    # If it's basically just a URL, no title guess
    if re.fullmatch(r"https?://\S+", s):
        return None

    # Pattern A: (YEAR). Title. Source...
    m = re.search(r"\(\s*(19|20)\d{2}\s*\)\.\s*(.+)", s)
    if m:
        tail = m.group(2).strip()
        # title is usually up to next period before journal/source/doi/url
        title = tail.split(". ")[0].strip().rstrip(".")
        title = DOI_TOKEN_RE.sub("", title).strip()
        title = re.sub(r"\bhttps?://\S+\b", "", title).strip()
        if DOI_RE.search(title):
            return None
        return title if 6 <= len(title) <= 300 else None

    # Pattern B: Authors. Title. Journal...
    # Find first "sentence" after authors chunk.
    # Heuristic: split on first period that is followed by a space and a capital letter
    parts = re.split(r"\.\s+(?=[A-Z])", s, maxsplit=2)
    if len(parts) >= 2:
        # parts[0] likely authors; parts[1] likely title or title+stuff
        title = parts[1].strip().rstrip(".")
        title = DOI_TOKEN_RE.sub("", title).strip()
        title = re.sub(r"\bhttps?://\S+\b", "", title).strip()
        if DOI_RE.search(title):
            return None
        return title if 6 <= len(title) <= 300 else None

    return None


# ----------------------------
# 4) Ollama enrichment (optional)
# ----------------------------

def ask_ollama_json(
    model: str,
    prompt: str,
    temperature: float = 0.0,
    host: str = "http://localhost:11434",
    timeout_s: int = 60,
):
    """
    Calls Ollama generate endpoint, streams response, and robustly extracts JSON.
    """
    r = requests.post(
        f"{host}/api/generate",
        json={"model": model, "prompt": prompt, "options": {"temperature": temperature}},
        stream=True,
        timeout=timeout_s
    )
    r.raise_for_status()

    out = ""
    for line in r.iter_lines():
        if not line:
            continue
        obj = json.loads(line.decode("utf-8"))
        out += obj.get("response", "")
        if obj.get("done", False):
            break

    # Robust JSON extraction: find first {...} blob
    # Handles cases where the model adds prose before/after JSON.
    m = re.search(r"\{.*\}", out, flags=re.DOTALL)
    if not m:
        return {}
    blob = m.group(0)

    try:
        return json.loads(blob)
    except json.JSONDecodeError:
        # last-ditch cleanup: remove trailing commas
        blob2 = re.sub(r",(\s*[}\]])", r"\1", blob)
        try:
            return json.loads(blob2)
        except Exception:
            return {}

def llm_parse_reference(model: str, raw: str, title_guess: Optional[str], doi: Optional[str], urls: List[str], year: Optional[int]) -> Dict[str, Any]:
    prompt = f"""
Return ONLY valid JSON. No commentary.

Task: refine a partially-parsed bibliography entry.

Rules:
- If title_guess is not null, set "title" exactly equal to title_guess unless you are highly confident it is wrong.
- For journal_or_source: if the reference contains a journal/source name, COPY it verbatim from the reference text (do not rewrite).
- Do not invent fields that are not present. If you cannot find a substring in the reference that corresponds to the field, use null.
- "authors" may be a string or a list of strings.
- If the entry is clearly a web resource (URL present, no DOI, no journal), set journal_or_source to the site or organization name if present; otherwise null.

Output JSON:
{{
  "authors": null,
  "title": null,
  "journal_or_source": null,
  "year": null,
  "volume": null,
  "issue": null,
  "pages": null,
  "doi": null,
  "url": null
}}

Known signals:
title_guess: {json.dumps(title_guess)}
doi_detected: {json.dumps(doi)}
urls_detected: {json.dumps(urls)}
year_detected: {json.dumps(year)}

Reference:
{raw}
""".strip()

    data = ask_ollama_json(model=model, prompt=prompt, temperature=0.0)
    return {
        "authors": data.get("authors"),
        "title": data.get("title"),
        "journal_or_source": data.get("journal_or_source"),
        "year_llm": data.get("year"),
        "volume": data.get("volume"),
        "issue": data.get("issue"),
        "pages": data.get("pages"),
        "doi_llm": data.get("doi"),
        "url_llm": data.get("url"),
    }


# ----------------------------
# 5) End-to-end: DOCX -> DataFrame
# ----------------------------

def _ensure_columns(df: pd.DataFrame, cols: List[str]) -> pd.DataFrame:
    for c in cols:
        if c not in df.columns:
            df[c] = pd.NA
    return df


def bibliography_docx_to_dataframe(
    docx_path: str,
    use_ollama: bool = False,
    model: str = "mistral",
    debug: bool = True,
) -> pd.DataFrame:
    paras = load_docx_paragraphs(docx_path)
    block = find_bibliography_block(paras)

    if debug:
        print(f"[debug] paragraphs total: {len(paras)}")
        print(f"[debug] bibliography block paragraphs found: {len(block)}")
        if len(block) > 0:
            print("[debug] first 3 block paragraphs:")
            for x in block[:3]:
                print("   -", x[:120], "..." if len(x) > 120 else "")

    refs = split_references_from_block(block)

    if debug:
        print(f"[debug] references detected: {len(refs)}")
        if len(refs) > 0:
            print("[debug] first reference raw (truncated):")
            print("   ", refs[0]['raw'][:200], "..." if len(refs[0]['raw']) > 200 else "")


    rows = []

    for r in refs:
        raw = r.get("raw", "")

    # ---- deterministic extraction FIRST ----
        row = {
            "ref_id": r.get("ref_id"),
            "raw": raw,
            "year": extract_year(raw),
            "doi": extract_doi(raw),
            "urls": extract_urls(raw),
            "title_guess": heuristic_title(raw),
        }

    # ---- LLM enrichment SECOND (optional) ----
        if use_ollama:
            row.update(
                llm_parse_reference(
                    model=model,
                    raw=raw,
                    title_guess=row["title_guess"],
                    doi=row["doi"],
                    urls=row["urls"],
                    year=row["year"] if pd.notna(row["year"]) else None,
                )
            )

    # ---- commit row ----
        rows.append(row)

    df = pd.DataFrame(rows)

    # Ensure expected columns exist even if df is empty
    EXPECTED_COLS = [
      "ref_id","raw","year","doi","urls","title_guess",
      "authors","title","journal_or_source","year_llm","volume","issue","pages",
      "doi_llm","url_llm"
    ]
    df = _ensure_columns(df, EXPECTED_COLS)


    # Flags (safe now)
    df["ref_id_is_missing"] = df["ref_id"].isna()
    df["ref_id_is_duplicate"] = df["ref_id"].duplicated(keep=False) & df["ref_id"].notna()

    return df

```

```python
def postprocess_bibliography_df(df: pd.DataFrame) -> pd.DataFrame:
    import pandas as pd

    def normalize_scalar(val):
        if val is None or (isinstance(val, float) and pd.isna(val)):
            return None
        if isinstance(val, (list, tuple, set)):
            parts = [str(x).strip() for x in val if str(x).strip()]
            return ", ".join(parts) if parts else None
        return str(val).strip()

    # normalize authors to a string
    if "authors" in df.columns:
        df["authors"] = df["authors"].apply(normalize_scalar)

    # Parquet safety for scalar fields that sometimes become lists
    for c in ["pages", "volume", "issue", "journal_or_source", "title", "doi_llm", "url_llm"]:
        if c in df.columns:
            df[c] = df[c].apply(normalize_scalar)

    # Prefer LLM title; fallback to title_guess
    if "title_final" not in df.columns:
        df["title_final"] = df.get("title")
    if "title" in df.columns and "title_guess" in df.columns:
        mask = df["title_final"].isna() | (df["title_final"].astype(str).str.strip() == "")
        df.loc[mask, "title_final"] = df.loc[mask, "title_guess"]

    return df
```

```python
%%time
file_path = Path("Introduction.docx")
paper_id = file_path.stem

df = bibliography_docx_to_dataframe(str(file_path), use_ollama=True, model="phi3")
df = postprocess_bibliography_df(df)
```

```python
def _norm(s: str) -> str:
    return re.sub(r"\s+", " ", (s or "")).strip()

def _strip_ref_id_prefix(s: str) -> str:
    return re.sub(r"^\[\s*\d+\s*\]\s*", "", s).strip()

def _strip_authors_prefix(raw_s: str, authors: Optional[str]) -> str:
    """Remove author string if it appears at the start, forgiving punctuation/spacing."""
    if not authors:
        return raw_s

    s = raw_s
    a = _norm(authors).rstrip(".")
    # Build a permissive regex where commas/periods/spaces are flexible
    a_re = re.escape(a)
    a_re = a_re.replace(r"\,", r"[,\s]*").replace(r"\.", r"[.\s]*")
    pat = re.compile(r"^\s*" + a_re + r"\s*[,.;:—-]*\s*", re.IGNORECASE)

    return pat.sub("", s, count=1).lstrip()

def _find_journal_pos_by_tokens(s: str, journal: Optional[str]) -> Optional[int]:
    """
    Find approximate start position of the journal/source by searching for distinctive tokens
    from journal name in the raw string (handles abbreviations like 'Res Lett').
    """
    if not journal:
        return None

    s_low = s.lower()
    tokens = [t.lower() for t in re.findall(r"[A-Za-z]{4,}", journal)]  # 4+ letters tokens
    if not tokens:
        return None

    # Find earliest occurrence of any token, prefer longer tokens
    tokens = sorted(set(tokens), key=len, reverse=True)
    positions = []
    for t in tokens[:6]:  # limit
        p = s_low.find(t)
        if p != -1:
            positions.append(p)

    return min(positions) if positions else None

def title_from_raw(raw: str, authors: Optional[str], journal: Optional[str]) -> Optional[str]:
    if not raw:
        return None

    s = _norm(raw)
    s = _strip_ref_id_prefix(s)

    # If URL-only, don't try to invent a title
    if URL_RE.fullmatch(s):
        return None

    # Remove authors prefix if possible
    s2 = _strip_authors_prefix(s, authors)

    # If we successfully stripped authors, title likely starts immediately
    # Otherwise, fallback: split after the first ". " (common "Authors. Title" format)
    if s2 != s:
        start_text = s2
    else:
        parts = s.split(". ", 1)
        start_text = parts[1] if len(parts) == 2 else s

    start_text = start_text.strip(" ,.;:-—")

    # Now decide where title ends.
    cut_positions = []

    # 1) If we can find journal approximately, cut before it
    jpos = _find_journal_pos_by_tokens(start_text, journal)
    if jpos is not None and jpos > 0:
        cut_positions.append(jpos)

    # 2) Cut before DOI / URL markers if present
    m = DOI_RE.search(start_text)
    if m:
        cut_positions.append(m.start())
    m = URL_RE.search(start_text)
    if m:
        cut_positions.append(m.start())

    # 3) Cut before a year token if it appears (often "2014;9:57" etc.)
    m = YEAR_RE.search(start_text)
    if m:
        cut_positions.append(m.start())

    # Choose earliest cut if any
    title_chunk = start_text if not cut_positions else start_text[:min(cut_positions)]
    title_chunk = title_chunk.strip(" .;,-")

    # Guardrails: reject if it still starts with authors
    if authors and title_chunk.lower().startswith(_norm(authors).lower()[:12]):
        # last-ditch: remove again
        title_chunk = _strip_authors_prefix(title_chunk, authors).strip(" .;,-")

    if len(title_chunk) < 8 or len(title_chunk) > 300:
        return None

    # Avoid returning "doi: ..."
    if DOI_TOKEN_RE.match(title_chunk) or DOI_RE.search(title_chunk):
        return None

    return title_chunk

```

```python
def is_present(x) -> bool:
    if x is None:
        return False
    if isinstance(x, float) and pd.isna(x):
        return False
    if isinstance(x, str) and x.strip() == "":
        return False
    return True

mask_need_title = (
    df["authors"].apply(is_present) &
    df["journal_or_source"].apply(is_present) &
    ~df["title_final"].apply(is_present)
)

for i in df.index[mask_need_title]:
    df.at[i, "title_final"] = title_from_raw(
        raw=df.at[i, "raw"],
        authors=df.at[i, "authors"],
        journal=df.at[i, "journal_or_source"]
    )

# Also fix cases where title_final accidentally begins with authors (like your ref 4)
mask_title_starts_with_authors = (
    df["authors"].apply(is_present) &
    df["title_final"].apply(is_present) &
    df.apply(lambda r: str(r["title_final"]).lower().startswith(str(r["authors"]).lower()[:10]), axis=1)
)

for i in df.index[mask_title_starts_with_authors]:
    df.at[i, "title_final"] = title_from_raw(
        raw=df.at[i, "raw"],
        authors=df.at[i, "authors"],
        journal=df.at[i, "journal_or_source"]
    )

df.loc[:, ["ref_id", "authors", "title_final", "journal_or_source"]]

```

```python
def journal_from_raw(raw: str) -> Optional[str]:
    if not raw:
        return None
    s = " ".join(raw.split())
    s = re.sub(r"^\[\s*\d+\s*\]\s*", "", s).strip()

    m = JOURNAL_YEAR_RE.search(s)
    if not m:
        return None

    journal = m.group(1).strip()
    # clean trailing junk
    journal = journal.strip(" ,;:-")
    return journal if 2 <= len(journal) <= 120 else None
```

```python
mask_need_journal = ~df["journal_or_source"].apply(is_present)

df.loc[mask_need_journal, "journal_or_source"] = df.loc[mask_need_journal, "raw"].apply(journal_from_raw)
```

```python
def normalize_scalar(val):
    """Convert list/tuple/set to a single string; keep scalars; preserve None/NaN."""
    if val is None or (isinstance(val, float) and pd.isna(val)):
        return None
    if isinstance(val, (list, tuple, set)):
        # join non-empty parts
        parts = [str(x).strip() for x in val if str(x).strip()]
        return ", ".join(parts) if parts else None
    return str(val).strip()

# columns that should be scalar strings/numbers (not lists)
scalar_cols = ["pages", "volume", "issue", "journal_or_source", "title", "title_final", "doi_llm", "url_llm"]

for c in scalar_cols:
    if c in df.columns:
        df[c] = df[c].apply(normalize_scalar)
```
